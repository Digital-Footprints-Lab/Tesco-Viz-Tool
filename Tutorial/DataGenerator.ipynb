Just generates and saves some dummy synthetic data to visualise in the app

```python
import numpy as np
import pandas as pd
```

```python
# Let's say we go shopping once a month
dates = [x for x in range(1, 13)]
num = len(dates)
dates
```

```python
Products = {"Weetabix": 3.50, "Milk": 1.45, "Bread": 1.95, "Eggs":2.65, "Cheese":4.00, "Apples":2.20, "Oranges":2.10}
Products
```

```python
ShoppingTrips = pd.DataFrame()
# Randomly generate the number of each product bought on each shopping trip
np.random.seed(0)
ShoppingTrips["Month"] = dates
for product in Products.keys():
    ShoppingTrips[product] = np.random.randint(0, 5, num)
ShoppingTrips
```

```python
# Month, Product, Number
TripsMelt = pd.melt(ShoppingTrips, id_vars=["Month"], var_name="Product", value_name="Number")
TripsMelt.head()
```

```python
TripsMelt[TripsMelt["Month"] == 1]
```

```python
import plotly.express as px
fig = px.bar(TripsMelt, x="Product", y = "Number", animation_frame="Month")
fig.show()
```

```python
ShoppingTrips.to_csv("ShoppingTrips.csv", index=0)
TripsMelt.to_csv("TripsMelt.csv")
```

```python
ShoppingCost = pd.DataFrame()
ShoppingCost["Month"] = ShoppingTrips["Month"]
for column in Products.keys():
    ShoppingCost[column] = ShoppingTrips[column] * Products[column]

ShoppingCost["Total"] = ShoppingCost.sum(axis=1)
ShoppingCost
```

```python
CostMelt = pd.melt(ShoppingCost, id_vars=["Month"], var_name="Product", value_name="Cost")
CostMelt.head()
```

```python
fig2 = px.bar(CostMelt, x="Product", y = "Cost", animation_frame="Month")
fig2.show()
```

```python
CostMelt.to_csv("CostMelt.csv")
```

```python
ShoppingCost.to_csv("ShoppingCost.csv", index=0)
```
